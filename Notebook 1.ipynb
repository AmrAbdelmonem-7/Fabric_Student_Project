```sql
%%pyspark
import requests
import json
url = "https://jsonplaceholder.typicode.com/users"
response = requests.get(url, timeout=10)
data = response.json()

rdd = spark.sparkContext.parallelize([json.dumps(user) for user in data])
api_df = spark.read.json(rdd)

api_df.select("id", "name", "email", "phone", "website") \
      .write.format("delta").mode("overwrite").saveAsTable("dbo.external_cloud_users")


display(spark.sql("SELECT * FROM dbo.external_cloud_users LIMIT 5"))
import pandas as pd

github_url = "https://raw.githubusercontent.com/datasets/gdp/master/data/gdp.csv"
pdf = pd.read_csv(github_url)

external_df = spark.createDataFrame(pdf.head(100))
external_df.write.format("delta").mode("overwrite").saveAsTable("dbo.external_github_gdp")

display(spark.sql("SELECT * FROM dbo.external_github_gdp LIMIT 5"))
```
